# 🔀 EidCloud AI Model Router Quickstart

Welcome to **eidcloud-model-router**! This notebook demonstrates how to:
1. Initialize the pure PHP router and register models across multiple AI providers.
2. Feed diverse incoming tasks (Code, Vision, Arabic, JSON) through the heuristic classifier.
3. Inspect dynamic model selections, constraint enforcement, and cost optimization savings in USD.

```python
# Cell 1: Setup environment and verify PHP router binary
!apt-get update -qq && apt-get install -y -qq php-cli
!php -v
!php bin/eidcloud-router --help
```

```python
# Cell 2: Feed diverse tasks and observe intelligent capability routing
tasks = [
    "Write a PHP 8.4 script implementing an LRU cache with strict typing",
    "Extract data from this uploaded invoice diagram and OCR receipt",
    "لخص المقال واشرح أهم النقاط الرئيسية باللغة العربية الفصحى",
    "Format this customer dataset strictly as valid json schema"
]

import subprocess
import json

print("=" * 75)
print(f"{'PROMPT SNIPPET':<30} | {'CATEGORY':<8} | {'SELECTED MODEL':<20} | {'EST. COST'}")
print("=" * 75)

for prompt in tasks:
    res = subprocess.run(
        ["php", "bin/eidcloud-router", "route", prompt, "--json"],
        capture_output=True, text=True
    )
    data = json.loads(res.stdout)
    cat = data['classification']['category'].upper()
    model = data['selected_model']['id']
    cost = f"${data['cost_estimation']['total_cost_usd']:.6f}"
    snippet = prompt[:27] + "..." if len(prompt) > 30 else prompt
    print(f"{snippet:<30} | {cat:<8} | {model:<20} | {cost}")
```
